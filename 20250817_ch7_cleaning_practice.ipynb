```python
# Q8 applyで派生列作成（df1を使用）
# データ: /mnt/data/20250817_df1.csv
# height_cm, weight_kg から BMI（kg/m^2）を計算して列 bmi を作成し、
# さらに bmi を基に「<18.5: 'Under', 18.5-25: 'Normal', >=25: 'Over'」のクラス列 bmi_cls を作成してください。
import pandas as pd
import numpy as np
df = pd.read_csv('20250817_df1.csv')
# ここに解答
#身長と体重がNaNの場合は中央値を入れる
df["height_cm"] = df["height_cm"].fillna(df["height_cm"].median())
df["weight_kg"] = df["weight_kg"].fillna(df["weight_kg"].median())

df["bmi"] = df.apply(lambda x:x["weight_kg"] / (x["height_cm"]*2)*100,axis=1)


#df["bmi"] = df["weight_kg"]/(df["height_cm"] * 2)*100
#print(df)

bins = [0,18.5,25,100]
group_names = ["Under","Normal","Over"]
bmi_cut = pd.cut(df["bmi"],bins, labels=group_names,right=False)
#print(bmi_cut)
df["bmi_cls"] = bmi_cut


print(df)
```

```python
# Q9 クリーニング規則の適用（salesを使用）
# データ: /mnt/data/20250817_sales.csv
# 以下のクリーニングを行い、クレンジング後の件数と合計売上(qty*price)を確認してください。
# 1) qty < 0 を 0 にクリップする
# 2) price <= 0 または price が欠損の行を除去
# 3) qty の欠損は 0 補完
# 4) 型を qty=int, price=float に整える
import pandas as pd
df = pd.read_csv('20250817_sales.csv', parse_dates=['date'])
# ここに解答


df["qty"] = df["qty"].fillna(0)
df["qty"] = df["qty"].clip(lower=0)

df[df["price"] <= 0] = np.nan
df = df.dropna(subset="price")

df["qty"] = pd.to_numeric(df["qty"], errors="coerce").astype(int)
df["price"] = pd.to_numeric(df["price"], errors="coerce").astype(float)

print(df)
```

```python
# Q10 インデックス操作（salesを使用）
# データ: /mnt/data/20250817_sales.csv
# store と date の複合インデックスに設定→日付昇順に並べ替え→最後に reset_index() で戻してください。
import pandas as pd
df = pd.read_csv('20250817_sales.csv', parse_dates=['date'])
# ここに解答
print(df)
df = df.set_index(["store","date"])
df = df.sort_index(level="date")
df = df.reset_index()

print(df)
```
