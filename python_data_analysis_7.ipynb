```python
import pandas as pd
import numpy as np

#欠損値を取り除く
data = pd.Series([1,np.nan,3.5,np.nan,7])
#print(data)
#print(data.dropna())

#いくつ欠損値があるかによって取り除くかどうか決める
data = pd.DataFrame([[1.,6.5,3.],[1.,np.nan,np.nan],[np.nan,np.nan,np.nan],[np.nan,6.5,3.]])

#print("data\n",data)
#print("data.dropna()\n",data.dropna())
#print("data.dropna(how=all)\n",data.dropna(how="all"))

#列に対して行う場合
new_data = data.copy()
new_data[3] = np.nan
print("new_data\n",new_data)
#print("new_data.dropna(axis=columns, how=all)\n", new_data.dropna(axis="columns",how="all"))

#欠損値の数がある上限値未満のときだけ保持したい場合
print(new_data.dropna(thresh=1))

```

```python
#欠損値の穴埋め
test_data = new_data.copy()

print(test_data)
print(test_data.fillna(0))
#列ごとに異なる値で埋める
print(test_data.fillna({0:0,1:100,2:200,3:300}))
#直前の値をそのまま入れる（limitでいくつ置き換えるか指定できる）
print(test_data.fillna(method="ffill",limit=1))
```

```python
#データの変形
#重複の除去
data = pd.DataFrame({"k1":["one", "two"] * 3 + ["two"], "k2":[1,1,2,3,3,4,4]})
print(data)

#重複かどうかを判定する
#print(data.duplicated())
#重複の場合消す
print(data.drop_duplicates())
#重複した値の最後のやつを残す
print(data.drop_duplicates(subset=["k1","k2"],keep="last"))
#指定した一部の列のみ
print(data.drop_duplicates(subset=["k1"]))
```

```python
#関数やマッピングを用いたデータ変換

data = pd.DataFrame({"food":["bacon","pulled pork", "bacon","pastrami","corned beef", "bacon","pastrami","honey ham","nova lox"],
                     "ounces":[4,3,12,6,7.5,8,3,5,6]})

#加工品に対してどの動物が使われたかマッピングする
meat_to_animal = {"bacon":"pig","pulled pork":"pig","pastrami":"cow","corned beef":"cow","honey ham":"pig","nova lox":"salmon"}

data["animal"] = data["food"].map(meat_to_animal)

data
```

```python
#値の置き換え
#replaceを使うのがシンプルで柔軟性が高いらしい！
data = pd.Series([1.,-999.,2.,-999.,-1000.,3.])

#特定の値を欠損値に変える
#print(data.replace(-999,np.nan))
#複数の値の場合はこう！
#print(data.replace([-999,-1000], np.nan))
#値に対応する値の場合はこうでこう！
print(data.replace([-999,-1000],[np.nan,0]))
#これでも良い。というか、こっちのほうがわかりやすいし、引数として使えそう
print(data.replace({-999:np.nan,-1000:0}))
#これはどうだ！？
map_dict1 = {-999:np.nan,-1000:0}
print(data.replace(map_dict1))
```

```python
#文字列操作
#カンマ区切り文字列を分割するときはsplit
val = "a,b,   guido"
print(val.split(","))
#前後の空白を除去するのはstrip
val_test2 = [x.strip() for x in val.split(",")]
print(val_test2)
#連結するのは.join
print("::".join(val_test2))
#検索系は in（その中に存在するか） .index()（()の中の文字があれば1がくる？） .find()(indexと同じだが無い時は-1が返ってくる。)
print(val.index(","))
print(val.find("Z"))
#数を数えるのはcount
print(val.count(","))
```

```python
#正規表現
import re
text = "foo    bar\t baz   \tqux"

print(text)
text2 = re.split(r"\s+",text)

print(text2)
#コンパイルで定義しておくと楽ちんらしい。あとfindallはマッチした文字列を取り出す。
text2_comp = re.compile(r"\s+")
print(text2_comp.split(text))
print(text2_comp.findall(text))

#searchメソッドは最初にマッチした1つだけを返す。matchは文字列の先頭でマッチするかのみ。
text = "Dave dave@google.com Steve steve@steve@gmail.com Rob rob@gmail.com Ryan ryan@yahoo.com"
pattern = r"[A-Z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,4}"
regex = re.compile(pattern, flags=re.IGNORECASE)
print(regex.findall(text))
print(regex.search(text))
#subは文字列内にパターンが見つかった際にその部分を指定した別の文字列に置き換える。
print(regex.sub("REDACTED",text))

pattern = r"([A-Z0-9._%+-]+)@([A-Z0-9.-]+)\.([A-Z]{2,4})"
regex = re.compile(pattern, flags=re.IGNORECASE)
test_match = regex.match("wesm@bright.net")
print(test_match.groups())
#\1は1番目のグループにマッチしたやつ。\2は2番目のグループにマッチしたやつ
print(regex.sub(r"Username: \1, Domain: \2, Suffix: \3", text))
#文字列メソッドはたくさんあるから、こういう時はこう使うようにしっかりやる。
#正規表現は実務的に非常に使うので、型とかをマスターしておきたい。
#特に文字列系は必要になるだろう。
```

```python
#カテゴリ型データ
#処理速度やメモリ使用量が改善する！

values1 = pd.Series(["apple","orange","apple","apple"]*2)
#print(pd.unique(values1))
#print(pd.value_counts(values1))

#整数値の値テーブルを作っておいて、その後にディメジョンテーブルを使って、values1に戻す。
values2 = pd.Series([0,1,0,0]*2)
dim = pd.Series(["apple","orange"])

print(dim.take(values2))

#これを使えば、カテゴリ名を変更したり、すでにあるカテゴリの順序や位置を変えずに新たなカテゴリを追加できる。
#カテゴリ型にしたい場合は
cate = values1.astype("category")
#このカテゴリ型のカテゴリ情報を抽出する。
c = cate.array

#カテゴリ情報を取得するには。
print(c.categories)
print(c.codes)

#カテゴリを用いた計算
N = 10000000
labels = pd.Series(["foo","bar","baz", "qux"] * (N // 4))
categories = labels.astype("category")

print(labels.memory_usage(deep=True))
print(categories.memory_usage(deep=True))
#カテゴリ変数にするとGroupbyの演算は非常に高速にできる。
```
