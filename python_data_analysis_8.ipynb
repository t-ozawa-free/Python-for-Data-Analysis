```python
import pandas as pd
import numpy as np
data = pd.Series(np.random.uniform(size=9),
                 index=[["a","a","a","b","b","c","c","d","d"],
                        [1,2,3,1,3,1,2,2,3]])

print(data)
print(data.index)

print(data["b"])

print(data["b":"c"])
```

```python
data2 = data.unstack()

print(data2)
print(data)
print(data2.stack())
```

```python
frame = pd.DataFrame(np.arange(12).reshape((4,3)),
                     index=[["a","a","b","b"],[1,2,1,2]],
                     columns=[["Ohio","Ohio","Colorado"],
                              ["Green","Red","Green"]])

print(frame)

frame.index.names = ["key1","key2"]

print(frame)

frame.columns.names = ["state","color"]

print(frame)
```

```python
print(frame.sort_index(level=1).sort_index(level=0))

test_frame = frame.groupby(level="key2").sum()

print(test_frame)

test_frame2 = frame.groupby(level="color",axis=1).sum()

print(test_frame2)
```

```python
import pandas as pd
import numpy as np

# サンプルデータ作成
df = pd.DataFrame({
    "city": ["Tokyo", "Tokyo", "Osaka", "Osaka"],
    "year": [2020, 2021, 2020, 2021],
    "category": ["A", "A", "B", "B"],
    "sales": [100, 120, 90, 130]
})

# Q1: city, yearでMultiIndexを作成
df_multi = df.set_index(["city", "year"])
print(df_multi)

# Q2: category列をunstack()してピボットテーブル化
print(df_multi.set_index("category", append=True).unstack("category"))

# Q3: MultiIndexのnamesを変更して表示
df_multi.index.names = ["都市", "年度"]
print(df_multi)
```
