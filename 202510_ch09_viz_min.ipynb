# 第9章：可視化（最小テンプレ）
- 目的：集計→図→保存を**毎回同じ型**で出せるようにする

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
DATA = Path.cwd() / "assets"
sales = pd.read_csv(DATA / "sales_sample.csv")
sales['date'] = pd.to_datetime(sales['date'], errors='coerce')
# 月次売上の推移（全体）
monthly = (sales.dropna(subset=['date'])
                 .set_index('date')
                 .resample('M')['amount'].sum())
fig = plt.figure()
monthly.plot()
plt.title('Monthly Sales (All Stores)')
plt.xlabel('Month'); plt.ylabel('Amount')
out = Path('assets') / 'fig_monthly_sales.png'
fig.savefig(out, bbox_inches='tight')
out
```

```python
# 店舗別の月次売上（pivot→plot）
stores = (sales.dropna(subset=['date'])
               .set_index('date')
               .groupby('store')['amount']
               .resample('M').sum()
               .unstack('store'))
fig2 = plt.figure()
stores.plot(ax=plt.gca())
plt.title('Monthly Sales by Store')
plt.xlabel('Month'); plt.ylabel('Amount')
out2 = Path('assets') / 'fig_monthly_by_store.png'
fig2.savefig(out2, bbox_inches='tight')
out2
```
