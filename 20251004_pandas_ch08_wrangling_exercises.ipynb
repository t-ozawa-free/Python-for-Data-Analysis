```python
# === Q1: DataFrame の連結 (concat) ===
# 2つのDataFrame df1, df2 を縦方向（行方向, axis=0）に連結してください。
import pandas as pd

df1 = pd.DataFrame({"A":[1,2], "B":[3,4]})
df2 = pd.DataFrame({"A":[5,6], "B":[7,8]})

# 期待:
#    A  B
# 0  1  3
# 1  2  4
# 0  5  7
# 1  6  8

# ここにコード
concat_data1 = pd.concat([df1,df2],axis=0)

print(df1)
print(df2)

print(concat_data1)
```

```python
# === Q2: インデックスを揃えて横方向に連結 ===
# df1 と df2 を列方向（axis=1）に連結してください。
# 結果の列名を "A1","B1","A2","B2" としてください（列のMultiIndexは使わない）。
import pandas as pd

df1 = pd.DataFrame({"A":[1,2,3], "B":[4,5,6]}, index=[0,1,2])
df2 = pd.DataFrame({"A":[7,8,9], "B":[0,1,2]}, index=[0,1,2])

# 期待: 列名が A1,B1,A2,B2 の4列

# ここにコード
concat_data2 = pd.concat([df1,df2],axis=1)
concat_data2.columns = ["A1","B1","A2","B2"]

print(df1)
print(df2)

print(concat_data2)
```

```python
# === Q3: データベース風の結合 (merge: 内部結合 / inner) ===
# 顧客DataFrameと注文DataFrameを "customer_id" で内部結合してください。
import pandas as pd

customers = pd.DataFrame({
    "customer_id":[1,2,3],
    "name":["Alice","Bob","Charlie"]
})
orders = pd.DataFrame({
    "order_id":[100,101,102,103],
    "customer_id":[1,2,2,4],
    "amount":[250,400,300,100]
})

# 期待: customer_id が 1 と 2 の注文のみ残る

# ここにコード
print(customers)
print(orders)

cust_orders = pd.merge(customers,orders,on="customer_id", how="inner")

print(cust_orders)
```

```python
# === Q4: 左外部結合 (left) ===
# customers を左、orders を右にして左外部結合を行い、
# 顧客が存在するが注文がない場合（customer_id=3）も残してください。
import pandas as pd

customers = pd.DataFrame({
    "customer_id":[1,2,3],
    "name":["Alice","Bob","Charlie"]
})
orders = pd.DataFrame({
    "order_id":[100,101,102,103],
    "customer_id":[1,2,2,4],
    "amount":[250,400,300,100]
})

# 期待: customer_id=3 の行も含まれ、注文関連列は欠損になる

# ここにコード
print(customers)
print(orders)

cust_orders2 = pd.merge(customers,orders,on="customer_id",how="outer")

print(cust_orders2)
```

```python
# === Q5: ピボット (pivot) ===
# 売上データを店舗×日付の形にピボットしてください（値は amount）。
import pandas as pd

sales = pd.DataFrame({
    "store":["A","A","B","B"],
    "date":["2025-10-01","2025-10-02","2025-10-01","2025-10-02"],
    "amount":[100,150,200,250]
})

# 期待: index=date, columns=store, values=amount の表（もしくは index=store, columns=date でもOK）

# ここにコード
sales_pivot = sales.pivot(index="date",columns="store",values="amount")

print(sales_pivot)
```

```python
# === Q6: stack / unstack ===
# Q5で作成したピボット表を stack して縦持ち形式に戻してください。
# その後 unstack で再び横持ちに変換してください。
# ここにコード

stack_p = sales_pivot.stack()

print(stack_p)

ustack_p = stack_p.unstack()

print(ustack_p)
```

```python
# === Q7: melt ===
# sales DataFrame を melt を使って "store" を id_vars に指定し、
# 他の列を縦に展開してください。
import pandas as pd

sales = pd.DataFrame({
    "store":["A","A","B","B"],
    "date":["2025-10-01","2025-10-02","2025-10-01","2025-10-02"],
    "amount":[100,150,200,250]
})

# 期待: 列 'variable','value' を持つ縦長データ（date/amountが変数側）

# ここにコード
melted = pd.melt(sales, id_vars="store")

print(melted)
```

```python
# === Q8: 複数キーでの結合 ===
# 部署・社員テーブルを「部署ID(dept_id)」と「社員ID(emp_id)」の複合キーで結合してください。
import pandas as pd

dept = pd.DataFrame({
    "dept_id":[10,20],
    "dept_name":["営業","開発"]
})
emp = pd.DataFrame({
    "dept_id":[10,10,20,20],
    "emp_id":[1,2,1,2],
    "emp_name":["佐藤","田中","鈴木","高橋"]
})
salary = pd.DataFrame({
    "dept_id":[10,20,20],
    "emp_id":[1,1,2],
    "salary":[300,400,500]
})

# 期待: dept_id/emp_id で emp と salary を結合（不足は欠損）

# ここにコード
emp_salary = pd.merge(emp,salary, on=["dept_id","emp_id"],how="outer")

print(emp_salary)
```
